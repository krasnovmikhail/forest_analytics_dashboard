```python
import os
import pandas as pd


def load_datasets_to_dict(folder_path):
  datasets_dict = {}

  # Проверяем все файлы в указанной папке
  for filename in os.listdir(folder_path):
    file_path = os.path.join(folder_path, filename)

    # Пропускаем папки, если они там есть
    if os.path.isdir(file_path):
      continue

    # Определяем расширение и считываем файл в DataFrame
    if filename.endswith(".xlsx") or filename.endswith(".xls"):
      # считываем Excel (можно указать конкретный sheet_name, если нужно)
      df = pd.read_excel(file_path)
      # Убираем расширение из имени для красивого ключа в словаре
      key_name = os.path.splitext(filename)[0]
      datasets_dict[key_name] = df
      print(f"Загружен Excel: {filename} -> ключ: '{key_name}'")

  print(f"\nВсего загружено датафреймов: {len(datasets_dict)}")
  return datasets_dict



my_data = load_datasets_to_dict('../datasets')
```
